```python

missing_count = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage
})

missing_table = missing_table.sort_values(
    by="Missing Values",
    ascending=False
)

display(missing_table)
```

```python

import matplotlib.pyplot as plt

missing_count = df.isnull().sum()
missing_count = missing_count[missing_count > 0]

missing_count.plot(kind="bar", figsize=(12, 5))

plt.title("Missing Values in Vehicle Emissions Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```

```python

total_rows = len(df)

duplicate_percentage = (duplicate_count / total_rows) * 100

print("Total rows:", total_rows)
print("Duplicate rows:", duplicate_count)
print("Duplicate percentage:", round(duplicate_percentage, 2), "%")
```

```python

fuel_emissions = df.groupby("Fuel type")["CO2 emissions (g/km)"].mean()

print(fuel_emissions)
```

```python

fuel_emissions.plot(kind="bar", figsize=(10, 5))

plt.title("Average CO2 Emissions by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average CO2 Emissions (g/km)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.scatter(
    df["Engine size (L)"],
    df["CO2 emissions (g/km)"],
    alpha=0.5
)

plt.title("Engine Size vs CO2 Emissions")
plt.xlabel("Engine Size (L)")
plt.ylabel("CO2 Emissions (g/km)")

plt.tight_layout()
plt.show()
```

```python

correlation = df["Engine size (L)"].corr(
    df["CO2 emissions (g/km)"]
)

print("Engine Size and CO2 Emissions Correlation:", round(correlation, 3))
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include="number")

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap - Vehicle Emissions Dataset")
plt.tight_layout()
plt.show()
```


# Final EDA Observations — GreenRoute

## 1. Vehicle Emission Dataset
- The dataset contains vehicle-related features and CO2 emission values.
- The CO2 emission column has no missing values.
- The average CO2 emission is approximately 253.71 g/km.
- CO2 emissions range from 94 to 608 g/km.
- Engine size and CO2 emissions have a positive correlation of approximately 0.828.
- Smog rating and CO2 rating contain missing values that may require handling in later stages.

## 2. Road Network Analysis
- The road network represents drivable roads in Gandhinagar, Gujarat.
- The network contains approximately 9,895 intersections and 26,522 road segments.
- Road length and road type analysis help describe the structure of the road network.

## 3. Elevation Analysis
- Elevation values were extracted from the downloaded raster dataset for road intersections.
- Elevation varies across the mapped area.
- Elevation information can help investigate terrain-related differences between routes.

## 4. Slope Analysis
- Elevation differences between connected intersections were used to estimate road-segment slopes.
- The slope distribution includes uphill, downhill, and approximately flat segments.
- Extreme slope values should be validated before using them for route optimization.

## 5. Route Comparison
- A shortest-distance route and an experimental slope-penalty route were compared.
- The slope-penalty route is an experimental approach, not a validated fuel-consumption model.
- CO2 estimates based on average emissions do not account for actual vehicle type, traffic, or road slope.

## Conclusion
The exploratory data analysis provides an initial understanding of vehicle emissions, road-network structure, elevation, slope, and route alternatives. These findings can guide future development of the GreenRoute project.
